```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
class GDRegressor:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.learning_rate = learning_rate
        self.epochs=epochs
        self.coef=None
        self.intercept=None

    def fit(self,X_train,y_train):
        self.intercept=0
        self.coef=np.ones(X_train.shape[1])

        for i in range(self.epochs):
            # clalcuale the prediction
            y_hat=np.dot(X_train,self.coef)+self.intercept
            
            # clacuale the intercept derivative
            intercept_der=-2 *np.mean(y_train-y_hat)
            # update the intercept
            self.intercept=self.intercept-(self.learning_rate*intercept_der)

            # calculate cofficent derivativr(all at once)
            coef_der = -2 / X_train.shape[0] * np.dot(X_train.T, (y_train - y_hat))

            #upadte the cofficienr

            self.coef=self.coef-(self.learning_rate*coef_der)
        print(f'Intercept:{self.intercept}')
        print(f'Corfficients:{self.coef}')

    def predict(self,X_test):
        return np.dot(X_test,self.coef)+self.intercept
```

```python
from sklearn.datasets import load_diabetes

# Load the dataset
X, y = load_diabetes(return_X_y=True)

print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Compare with sklearn's LinearRegression
sklearn_reg = LinearRegression()
sklearn_reg.fit(X_train, y_train)

print(f"\nSklearn Coefficients: {sklearn_reg.coef_}")
print(f"Sklearn Intercept: {sklearn_reg.intercept_}")

sklearn_pred = sklearn_reg.predict(X_test)
print(f"Sklearn R² Score: {r2_score(y_test, sklearn_pred)}")


# Our Gradient Descent Implementation
gd_reg = GDRegressor(learning_rate=0.5,epochs=1000)
gd_reg.fit(X_train, y_train)

gd_pred = gd_reg.predict(X_test)
print(f"\nGD Regressor R² Score: {r2_score(y_test, gd_pred)}")
```
