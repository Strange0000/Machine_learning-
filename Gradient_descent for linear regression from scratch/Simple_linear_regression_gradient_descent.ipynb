```python
import numpy as np
```

```python
class GDRegression:

    def __init__(self,learning_rate,epochs):
        self.learning_rate=learning_rate
        self.epochs=epochs
        self.m=None
        self.b=None
        self.loss_history=[]

    def fit(self,X,y):
        # initalize the parameter
        self.b=np.random.randn()
        self.m=np.random.randn()

        for epoch in range(self.epochs):
        # calculate the prediction
            y_pred=self.m+X+self.b

        # calculate the gradient
            slop_b=-2*(np.sum(y-self.m*X-self.b))
            slop_m=-2*(np.sum((y-self.m-self.b)*X))

        # update the parameters
            self.b=self.b-self.learning_rate*slop_b
            self.m=self.m-self.learning_rate*slop_m

        #calcualte n store Loss(MSE)
            loss=np.mean(y-self.predict(X)**2)
            self.loss_history.append(loss)

        return self

    def predict(self,X):
        return self.m*X+self.b

    def get_params(self):
        return{'Slope':self.m,'Intercept':self.b}
```

```python
GD=GDRegression(0.01,100)
```

```python
GD.fit(X,y)
```

```python
GD.predict(y)
```

```python
print(f'slop:{GD.m:.4f}')
print(f'intercept:{GD.b:.4f}')
```

```python
#plot the sloss
plt.plot(GD.loss_history)
plt.xlabel('Epochs')
plt.ylabel('loss (MSE)')
plt.title('Loss Curvr durning Traning')
```
