```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("train.csv")
```

```python
df.head()
```

```python
df['Fare'].plot(kind='kde')
```

```python
# sample size =30 - 10 times times

samples=[]
stds=[]

for i in range(10):
    x=df['Fare'].dropna().sample(30).values
    stds.append(x.std())
    samples.append(x.tolist())
```

```python
samples=np.array(samples)
```

```python
sampling_mean=samples.mean(axis=1)
```

```python
sample_Std=np.mean(stds)
```

```python
import seaborn as sns
```

```python
sampling_mean
```

```python
sns.kdeplot(sampling_mean)
```

```python
sample_Std
```

```python
lower_limit=sampling_mean.mean()-2.045*(sample_Std/np.sqrt(30))
upper_limit=sampling_mean.mean()+2.045*(sample_Std/np.sqrt(30))
```

```python
print("The range is",lower_limit,"_",upper_limit)
```
