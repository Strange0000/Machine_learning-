# Box plot 

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot of Age grouped by Survival status
sns.boxplot(data=data, x='Survived', y='Age')
plt.title('Age Distribution by Survival Status')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot combined with a stripplot to see individual data points
sns.boxplot(data=data, x='Survived', y='Age', color='lightblue')
sns.stripplot(data=data, x='Survived', y='Age', color='black', alpha=0.3, jitter=0.2)

plt.title('Age Distribution with Individual Data Points')
plt.show()
```

# Covariance - nature 

```python
df=pd.DataFrame()
```

```python
from re import X
x=pd.Series([12,25,68,42,113])
y=pd.Series([11,29,58,121,100])
```

```python
df['x']=x
df['y']=y
```

```python
df
```

```python
# plot on each exis

fig,(ax1,ax2)=plt.subplots(1,2,figsize=(10,3))
ax1.scatter(df['x'],df['y'])
ax2.scatter(df['x']*2,df['y']*2)

ax1.set_title("Covariance-"+str(np.cov(df['x'],df['y'])[0,1]))

ax2.set_title("Covariance-"+str(np.cov(df['x']*2,df['y']*2)[0,1]))
```

# Correlation -mangnitude`

```python
# plot on each exis

fig,(ax1,ax2)=plt.subplots(1,2,figsize=(10,3))
ax1.scatter(df['x'],df['x'])
ax2.scatter(df['x'],df['y'])

ax1.set_title("Correlation-"+ str(df['x'].corr(df['x'])))

ax2.set_title("Correlation-"+str(df['x'].corr(df['y'])))
```

```python
# plot on each exis

fig,(ax1,ax2)=plt.subplots(1,2,figsize=(10,3))
ax1.scatter(df['x'],df['y'])
ax2.scatter(df['x']*2,df['y']*2)

ax1.set_title("Correlation-"+ str(df['x'].corr(df['y'])))

ax2.set_title("Correlation-"+str((df['x']*2).corr(df['y']*2)))
```
