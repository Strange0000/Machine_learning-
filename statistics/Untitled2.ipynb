# Central Limit Theorem

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set of parameter
num_samples=10000
sample_size=300
distribution_range=(0,1)

# Generate sample from uniform distribution
sample=np.random.uniform(distribution_range[0],distribution_range[1],(num_samples,sample_size))

# CALCULATE THE SAMPLE MEAN
sample_mean=np.mean(sample,axis=1)

# Plot the histogram of thr sample mean
plt.hist(sample_mean,bins=30,density=True,edgecolor='black')
plt.title('Histogram of Sample Mean')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.show()
```

# CLT Case Study

```python
import numpy as np

# Set the parameters
population_size=100000
sample_size=50
num_sample=100

# Generate a random representative sample of salaries (in thousand)
# You should replace this with actual collected salary data
np.random.seed(42) # Setting a seed for reproducibility
population_Salaries=np.random.lognormal(mean=4.5,sigma=0.8,size=population_size)


```

```python
plt.hist(population_Salaries,bins=50,density=True,edgecolor='black')
```

```python
import seaborn as sns
sns.kdeplot(population_Salaries)
```

```python
#generate mutiple sample and calculate the sample means and standard deviations
sample_means=[]
sample_Std_dev=[]
for _ in range(num_samples):
    sample_salaries=np.random.choice(population_Salaries,size=sample_size)
    sample_means.append(np.mean(sample_salaries))
    sample_Std_dev.append(np.std(sample_salaries))
```

```python
# calculate the average of the sample means and the standard error
average_sample_means=np.mean(sample_means)
standard_error=np.std(sample_means)/np.sqrt(num_samples)
```

```python
# calculate the 95% condifence interval
margin_of_error = 1.96 * standard_error
lower_limit = average_sample_means - margin_of_error
upper_limit = average_sample_means + margin_of_error

# Report the results
print(f"Estimated average salary (in thousands): {average_sample_means:.2f}")
print(f"95% confidence interval (in thousands): ({lower_limit:.2f}, {upper_limit:.2f})")
```
